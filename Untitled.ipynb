```python
def write_nd_to_tif(nd, meta, transform, filename):
    
    meta.update({"driver": "GTiff",
                 "height": nd.shape[1],
                 "width": nd.shape[2],
                 "transform": transform})

    with rasterio.open(filename, "w", **meta) as dest:
        dest.write(nd)
```

```python
write_nd_to_tif(nd=lcm2015_london, 
                meta=lcm2015_new.meta, 
                transform=lcm2015_london_transform, 
                filename="lcm2015_london.tif")
```

### Reclassify Land Cover

```python
lcm2015_london2 = copy.deepcopy(lcm2015_london_np)

print(lcm2015_london2.shape)
```

```python
lookup = {1:1, 2:1,
          4:1, 5:1, 6:1, 7:1, 8:1, 9:1, 10:1, 11:1,
          3:2,
          12:2, 13:2, 14:2, 15:2, 16:2, 17:2, 18:2, 19:2,
          20:2, 21:2}
```

```python
for old, new in lookup.items(): 
    lcm2015_london2[lcm2015_london_np == old] = new

unique, counts = np.unique(lcm2015_london2, return_counts=True)  
counts2015_2 = dict(zip(unique, counts))                         
print(counts2015_2)
```

### Zonal Statistic

```python
lcm2015_london2_dr = rio.open("lcm2015_london2.tif")
lcm2015_london2_np = lcm2015_london2_dr.read(1)
```

```python
print(lcm2015_london2_np.shape)
```

```python
fig,ax=plt.subplots()
show((lcm2015_london2_dr,1), cmap=lcmap, norm=bnorm, ax=ax)
london.plot(ax=ax, facecolor='None', edgecolor='red', linewidth=0.3)
plt.show()
```

```python
zs_london15 = zonal_stats(london, lcm2015_london2_np,
                           affine=lcm2015_london2_dr.meta['transform'],
                           categorical=True)
zs_london15
```

```python
zs_london15_pd = pd.DataFrame(zs_london15)
zs_london15_pd.head()
```

```python
zs_london15_pd.set_index(london.index, inplace=True)

zs_london15_pd.rename(columns=idlabs, inplace=True) 

zs_london15_pd = zs_london15_pd.add_suffix('15')

zs_london15_pd = zs_london15_pd.fillna(0)

london_gpd = pd.merge(london, zs_london15_pd, how='left', left_index=True, right_index=True)
london_gpd.head()
```

```python
london_gpd['Sum15'] = london_gpd[['Green15', 'Non-Green15']].sum(axis=1) 

london_gpd['PctGreen'] = london_gpd['Green15'] / london_gpd['Sum15']
```

```python
np.random.seed(10)
moran_london = esda.moran.Moran(london_gpd['PctDegree'], londonQW)
```

```python
round(moran_london.I,3)
```

```python
moran_london.p_sim
```

```python
london_gpd['PctDegree_lag'] = weights.spatial_lag.lag_spatial(londonQW, london_gpd['PctDegree'])
```

```python
def standardize(df, var):
    
    newname = var + '_z'
    df[newname] = (df[var] - df[var].mean()) / df[var].std()
```

```python
standardize(london_gpd,'PctDegree')
standardize(london_gpd,'PctDegree_lag')
```

```python
lisa_london = esda.moran.Moran_Local(london_gpd['PctDegree'], londonQW, seed=10)
```

```python
esdaplot.lisa_cluster(lisa_london, london_gpd, p=1)
```

```python
esdaplot.lisa_cluster(lisa_london, london_gpd, p=0.05)
```

```python
lisa_london_fdr95 = esda.fdr(pvalues=lisa_london.p_sim, alpha=0.05)
lisa_london_fdr95
```

```python
esdaplot.lisa_cluster(lisa_london, london_gpd, p=lisa_london_fdr95)
```

```python
fig, axs = plt.subplots(1,2, figsize=(12, 12))

london_gpd.plot(ax=axs[0], column='PctDegree', cmap='Blues', 
        scheme='quantiles', legend=True,
       edgecolor="white", linewidth=0.1)
axs[0].set_axis_off()
axs[0].set_title("% Degree Holders")

esdaplot.lisa_cluster(lisa_london, london_gpd, p=lisa_london_fdr95, ax=axs[1])
axs[1].set_title("Local Moran's I (Queen) - % Degree Holders")

fig.tight_layout()
plt.savefig("Figure4_LISA_Degree.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
sig_mask = lisa_london.p_sim <= lisa_london_fdr95

london_gpd['LISA_q'] = 0 
london_gpd.loc[sig_mask, 'LISA_q'] = lisa_london.q[sig_mask]

cluster_labels = {0: 'Not Significant', 1: 'HH', 2: 'LH', 3: 'LL', 4: 'HL'}
london_gpd['LISA_cluster'] = london_gpd['LISA_q'].map(cluster_labels)

print(london_gpd['LISA_cluster'].value_counts())
```

### Figure 5

```python
hh_ll = london_gpd[london_gpd['LISA_cluster'].isin(['HH', 'LL'])].copy()

hh_data = hh_ll[hh_ll['LISA_cluster'] == 'HH']['PctGreen']
ll_data = hh_ll[hh_ll['LISA_cluster'] == 'LL']['PctGreen']

print(f"HH LSOAs (Elite Regime):    {len(hh_data)}")
print(f"LL LSOAs (Deprived Regime): {len(ll_data)}")
print()
print(hh_ll.groupby('LISA_cluster')['PctGreen'].describe())
```

```python
fig, ax = plt.subplots(1, figsize=(8, 6))

ax.boxplot(
    [hh_data, ll_data],
    labels=['HH (High Human Capital)', 'LL (Low Human Capital)'],
    patch_artist=True,
    boxprops=dict(facecolor='lightblue', color='black'),
    medianprops=dict(color='black', linewidth=2)
)

ax.set_ylabel('Proportion Green Land Cover (PctGreen)')
ax.set_title("% Green Land Cover by LISA Cluster Type\n"
             "(High-High vs Low-Low Degree Holder Regimes)")

plt.savefig("Figure5_Boxplot_GreenPremium.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
print(f"Mean PctGreen — HH (Elite Regime):    {hh_data.mean():.4f}")
print(f"Mean PctGreen — LL (Deprived Regime): {ll_data.mean():.4f}")
print(f"Green Premium (HH mean - LL mean):    {hh_data.mean() - ll_data.mean():.4f}")
```

### Figure 6

```python
borough_gpd = london_gpd.dissolve(
    by='LAD22NM',
    aggfunc={'PctDegree': 'mean', 'PctGreen': 'mean'}
).reset_index()

boroughQW = weights.contiguity.Queen.from_dataframe(
    borough_gpd, ids=borough_gpd['LAD22NM'].tolist()
)
boroughQW.transform = 'R'

np.random.seed(10)
moran_borough = esda.moran.Moran(borough_gpd['PctDegree'], boroughQW)
print(f"Borough Moran's I: {round(moran_borough.I, 3)}, p = {moran_borough.p_sim}")

lisa_borough = esda.moran.Moran_Local(borough_gpd['PctDegree'], boroughQW, seed=10)
```

```python
fig, axs = plt.subplots(1, 2, figsize=(14, 6))


axs[0].scatter(london_gpd['PctDegree'], london_gpd['PctGreen'],
               alpha=0.3, s=4, color='steelblue')
clean = london_gpd[['PctDegree','PctGreen']].dropna()
m, b = np.polyfit(clean['PctDegree'], clean['PctGreen'], 1)
x = np.linspace(clean['PctDegree'].min(), clean['PctDegree'].max(), 100)
axs[0].plot(x, m*x + b, color='red', linestyle='--', linewidth=1.5)
axs[0].set_xlabel('% Degree Holders')
axs[0].set_ylabel('% Green Cover')
axs[0].set_title(f'LSOA scale (n = {len(clean)})\nslope = {m:.3f}')


axs[1].scatter(borough_gpd['PctDegree'], borough_gpd['PctGreen'],
               alpha=0.8, s=60, color='darkblue', edgecolor='white')
clean_b = borough_gpd[['PctDegree','PctGreen']].dropna()
mb, bb = np.polyfit(clean_b['PctDegree'], clean_b['PctGreen'], 1)
x_b = np.linspace(clean_b['PctDegree'].min(), clean_b['PctDegree'].max(), 100)
axs[1].plot(x_b, mb*x_b + bb, color='red', linestyle='--', linewidth=1.5)
axs[1].set_xlabel('% Degree Holders')
axs[1].set_ylabel('% Green Cover')
axs[1].set_title(f'Borough scale (n = {len(clean_b)})\nslope = {mb:.3f}')

plt.suptitle('Human Capital vs Green Cover at Two Spatial Scales', fontsize=14)
plt.tight_layout()
plt.savefig('Figure6_MAUP_comparison.png', dpi=300, bbox_inches='tight')
plt.show()
```
